```python
word = input("Enter your text: ")
word = word.lower()
vowels = 0
consonants = 0

for letter in word:
  if letter in "aiueo":
    vowels += 1
  elif letter.isalpha():
    consonants += 1

characters = len(word)

print("Vowels: ", vowels)
print("Consonants: ", consonants)
print("Characters: ", characters)
```
